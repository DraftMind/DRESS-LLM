```python
from transformers import AutoModelForCausalLM, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("/home/ynliu/models/Qwen3-8B")
```

```python
messages =  [
    {"role": "user", "content": 'How are you? /no_think'},
    {"role": "assistant", "content": "<think>\n\n</think>\n\nI'm Fine."}
]

# 将messages转换为字符串格式

messages_string = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=False,   # 关键：让模型从 assistant 段落开始生成
    tokenize=False
)
```

```python
import os, glob, torch

model_dir = "/data1/liyu25/ynliu/DRESS-LLM/session/Qwen3_14B_Shakes/edited_model/seed_42_top_64_heads_alpha_1.5"  # 指向你保存的目录
wanted = ("self_attn.q_proj.bias", "self_attn.k_proj.bias",
          "self_attn.v_proj.bias", "self_attn.o_proj.bias")
found = 0

for f in glob.glob(os.path.join(model_dir, "pytorch_model*.bin")):
    sd = torch.load(f, map_location="cpu")
    for k, v in sd.items():
        if any(w in k for w in wanted):
            print(f"{os.path.basename(f)} | {k} | shape={tuple(v.shape)} | mean={float(v.mean())}")
            found += 1

print("total bias tensors found:", found)
# 另外也确认 config.json 里 "attention_bias": true
```
